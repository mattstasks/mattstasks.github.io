### Create an NLP pipeline

```python
# copy down the helper functions
def lower_replace(series):
    output = series.str.lower()
    output = output.str.replace(r'\[.*?\]', '', regex=True)
    output = output.str.replace(r'[^\w\s]', '', regex=True)
    return output

def token_lemma_nonstop(text):
    doc = nlp(text)
    output = [token.lemma_ for token in doc if not token.is_stop]
    output = ' '.join(output)
    return output

def filter_pos(text, pos_list=['NOUN', 'PROPN']):
    doc = nlp(text)
    output = [(token.text) for token in doc if token.pos_ in pos_list]
    output = ' '.join(output)
    return output
```

```python
# create an nlp pipeline
def nlp_pipeline(series):
    output = lower_replace(series)
    output = output.apply(token_lemma_nonstop)
    output = output.apply(filter_pos)
    return output
```

```python
import pandas as pd
import ast

df_genres = pd.read_csv("movies_info.csv")

# convert string representation of list into actual Python list
df_genres['genres'] = df_genres['genres'].apply(ast.literal_eval)
df_genres
```

```python
import pandas as pd
import ast
from sklearn.preprocessing import MultiLabelBinarizer


# Apply MultiLabelBinarizer
#genres for MultiLabelBinarizer (categorical features).
mlb = MultiLabelBinarizer()
genre_matrix = mlb.fit_transform(df_genres['genres'])

# Create a new DataFrame with binary columns
df_genres = pd.DataFrame(genre_matrix, columns=mlb.classes_, index=df_genres.index)

df_genres
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words='english')
overview_matrix = tfidf.fit_transform(df_clean['overview'])
overview_matrix
```

###Combine overview + genre features By adding genre encoding:

Drama + Crime movies are pulled closer together
Action + Fantasy movies are pulled closer together

```python
from scipy.sparse import hstack, csr_matrix

genre_sparse = csr_matrix(df_genres)
combined_features = hstack([overview_matrix, genre_sparse])
combined_features
```

###Compute similarity

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(combined_features)
```

```python
df1 = pd.read_csv("movies_info.csv")
df1.columns
```

```python
df1
```

```python
def recommend_movies(title, df1, similarity_matrix, top_n=10):
    # Find the index of the movie
    idx = df1[df1['original_title'].str.lower() == title.lower()].index
    if len(idx) == 0:
        return f"Movie '{title}' not found in dataset."
    idx = idx[0]

    # Get similarity scores for that movie
    sim_scores = list(enumerate(similarity_matrix[idx]))

    # Sort by similarity score (highest first)
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Exclude the movie itself
    sim_scores = sim_scores[1:top_n+1]

    # Extract indices and scores
    movie_indices = [i[0] for i in sim_scores]
    scores = [i[1] for i in sim_scores]

    # Build result DataFrame with scores
    result = df1[['original_title', 'genres']].iloc[movie_indices].copy()
    result['similarity_score'] = scores

    return result

# Example usage
print(recommend_movies("Blood & Gold", df1, similarity_matrix, top_n=10))
```

###2. Prepare the text columns

```python
# Apply pipeline to the combined_text column
movies["combined_text"] = nlp_pipeline(movies["combined_text"])

```

###3. Load sentence-transformers model

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer("all-MiniLM-L6-v2")
#Size: Small (22MB)
#Speed: Very fast, suitable for large datasets
#Accuracy: Strong performance on semantic similarity
#Use case: Great balance of speed and quality for movie recommendations
```

###4. Create embeddings
This converts every movie into a numeric embedding vector.

```python
embedding_matrix = model.encode(
    movies["combined_text"].tolist(),
    show_progress_bar=True
)
```

```python
import os
import numpy as np

if os.path.exists('embedding_matrix.npy'):
    embedding_matrix = np.load('embedding_matrix.npy')
else:
    embedding_matrix = model.encode(movies["combined_text"].tolist(), show_progress_bar=True)
    np.save('embedding_matrix.npy', embedding_matrix)
```

```python
#To import embeddings for future use without wasting time to run again
import numpy as np

# Load embeddings
embedding_matrix = np.load('embedding_matrix.npy')

# Now you can use embedding_matrix in your recommendation functions
```

###5. Create recommendation function

```python
def recommend_movies_2(movie_title, movies, embedding_matrix, top_n=10):
    # case-insensitive title matching
    title_to_index = {title.lower(): idx for idx, title in enumerate(movies["original_title"])}

    if movie_title.lower() not in title_to_index:
        return f"Movie title '{movie_title}' not found in the dataset."

    movie_idx = title_to_index[movie_title.lower()]

    # get embedding of selected movie
    movie_vector = embedding_matrix[movie_idx].reshape(1, -1)

    # compute cosine similarity
    sim_scores = cosine_similarity(movie_vector, embedding_matrix).flatten()

    # exclude the input movie itself
    sim_scores[movie_idx] = -1

    # get top N similar movies
    top_indices = sim_scores.argsort()[::-1][:top_n]

    # create output dataframe
    recommendations_2 = movies.iloc[top_indices][["original_title", "genres"]].copy()
    recommendations_2["similarity_score"] = sim_scores[top_indices]

    return recommendations_2
```

```python
recommend_movies_2("Blood & Gold", movies, embedding_matrix, top_n=10)

```

###6. Create an interactive function for embeddings

```python
movie_input = input("Enter a movie title: ")

# normalize input: lowercase + remove spaces
normalized_input = movie_input.lower()

results = recommend_movies_2(normalized_input, movies, embedding_matrix, top_n=10)

if isinstance(results, str):
    print(results)
else:
    print(f"\nTop 10 recommendations for '{movie_input}':\n")
results

```
